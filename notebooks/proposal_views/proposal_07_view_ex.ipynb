```python
# 2. 필요한 모듈 임포트
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import display

from services.data_preparer import prepare_proposal_07_data
from services.proposal_views.proposal_07_view import create_figure_and_df

# --- 테스트 실행 ---

# 1. app.py의 역할 (1): 데이터 준비 함수를 호출하여 필요한 데이터를 미리 로드합니다.
print("Step 1: `data_preparer`를 통해 분석용 데이터와 순서 정보를 준비합니다...")

filtering = '남성' # prepare_proposal_07_data() 함수에 삽입
# 글로벌 필터는 모두 기본값('전체')으로 호출
data_bundle = prepare_proposal_07_data()
analysis_df = data_bundle.get("analysis_df", pd.DataFrame())
order_map = data_bundle.get("order_map", {})
print(" -> 데이터 준비 완료!")


# 2. app.py의 역할 (2): 사용자가 Streamlit 위젯을 통해 필터를 선택했다고 가정합니다.
# 이 분석은 X축(첫 직무)과 그룹(경력 유형)이 고정되어 있고,
# 글로벌 필터는 data_preparer 단계에서 이미 적용되었습니다.
# proposal_01, 03과 달리 view 단계에서는 추가적인 드릴다운 선택이 없습니다.
# 하지만, view가 dimension_ui_name을 필요로 하므로 시뮬레이션 해줍니다.

# ----- 테스트하고 싶은 값으로 변경 -----
# 이 그래프는 '부서별' 또는 '직위직급별'로 필터링된 결과를 요약 테이블에 표시합니다.
# 그래프 자체는 동일하게 보이지만, 테이블의 컬럼이 바뀝니다.
selected_dimension_ui = '성별' # '부서별', '직위직급별'
# -----------------------------------

print(f"Step 2: 사용자가 '{selected_dimension_ui}' 기준으로 요약 테이블 보기를 선택했습니다.")


# 3. app.py의 역할 (3): view 함수에 준비된 모든 데이터와 필터 값을 전달하여 결과물 생성
print("Step 3: `view` 모듈을 호출하여 그래프와 요약 테이블을 생성합니다...")
if not analysis_df.empty:
    # 이 뷰는 드릴다운 기능이 없으므로 drilldown_selection은 '전체'로 고정합니다.
    # dimension_config는 이 뷰에서 직접 사용하지 않으므로 빈 딕셔너리를 전달해도 무방합니다.
    fig, aggregate_df = create_figure_and_df(
        analysis_df=analysis_df, 
        dimension_ui_name=selected_dimension_ui, 
        #drilldown_selection='전체',
        #dimension_config={}, # 이 view에서는 사용하지 않음
        order_map=order_map 
    )
    print(" -> 생성 완료!")
else:
    print(" -> 분석할 데이터가 없어 빈 결과물을 생성합니다.")
    fig, aggregate_df = go.Figure(), pd.DataFrame()


# --- 결과 확인 ---

# 4. ipynb에서 생성된 그래프를 확인합니다.
print("\n--- [결과 1] 생성된 Plotly 그래프 ---")
# pio.renderers.default = 'vscode' 
fig.show()

# 5. ipynb에서 생성된 요약 테이블(aggregate_df)을 확인합니다.
print(f"\n--- [결과 2] '{selected_dimension_ui}' 기준 생성된 요약 테이블 ---")
display(aggregate_df)
```
