```python
# 2. 필요한 모듈 임포트
import pandas as pd
import plotly.graph_objects as go
import plotly.io as pio
from IPython.display import display

from services.data_preparer import prepare_proposal_06_data
from services.proposal_views.proposal_06_view import create_figure

# --- 테스트 실행 ---

# 1. app.py의 역할 (1): 데이터 준비 함수를 호출하여 필요한 모든 데이터를 미리 로드합니다.
print("Step 1: `data_preparer`를 통해 모든 차원의 코호트 데이터를 준비합니다...")
data_bundle = prepare_proposal_06_data() # 글로벌 필터는 모두 기본값('전체')으로 호출
cohort_data_bundle = data_bundle.get("cohort_data_bundle", {})
order_map = data_bundle.get("order_map", {})
print(" -> 데이터 준비 완료!")


# 2. app.py의 역할 (2): 사용자가 Streamlit 위젯을 통해 필터를 선택했다고 가정합니다.
# 이 값들을 바꾸면서 여러 경우를 테스트해볼 수 있습니다.

# 2-1. 사용자 선택 시뮬레이션
# ----- 테스트하고 싶은 값으로 변경 -----
selected_dimension_ui = '성별'  # '부서별', '직무별', '직위직급별', '성별' 등 선택 가능
subgroup_name = '남성'          # 위 차원에 맞는 하위 그룹 선택 (예: 'Development Division', 'IT', 'Staff')
                                # 전체 그룹을 보려면 '전체' 입력
# -----------------------------------

print(f"Step 2: 사용자가 '{selected_dimension_ui}' 차원의 '{subgroup_name}' 그룹을 선택했습니다.")


# 3. app.py의 역할 (3): view 함수에 전달할 최종 데이터 '선택'
print("Step 3: `view` 모듈에 전달할 최종 데이터를 선택합니다...")
# 선택된 차원의 데이터 맵을 가져옴
cohort_map = cohort_data_bundle.get(selected_dimension_ui, {})
# 선택된 하위 그룹의 최종 코호트 피벗 테이블을 선택
cohort_pivot_to_plot = cohort_map.get(subgroup_name, pd.DataFrame())
title = f"[{selected_dimension_ui} - {subgroup_name}] 입사 연도별 잔존율"
print(" -> 데이터 선택 완료!")


# 4. view 함수 호출
print("Step 4: `view` 모듈을 호출하여 그래프를 생성합니다...")
if cohort_pivot_to_plot is not None and not cohort_pivot_to_plot.empty:
    # 이 view 함수는 요약 테이블을 반환하지 않으므로, aggregate_df는 _ 변수로 받습니다.
    fig, _ = create_figure(
        cohort_pivot=cohort_pivot_to_plot, 
        title=title
    )
    print(" -> 생성 완료!")
else:
    print(" -> 분석할 데이터가 없어 빈 그래프를 생성합니다.")
    fig, _ = go.Figure(), pd.DataFrame()


# --- 결과 확인 ---

# 5. ipynb에서 생성된 그래프를 확인합니다.
print("\n--- [결과] 생성된 Plotly 그래프 ---")
# pio.renderers.default = 'vscode' 
fig.show()
```
